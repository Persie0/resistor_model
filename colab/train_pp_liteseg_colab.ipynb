# Train PP-LiteSeg-T whole-resistor segmentation in Google Colab (Apache-2.0 path)

Use a GPU runtime, then choose **Runtime → Run all**. The script mounts Google Drive. No Ultralytics code, models or weights are used anywhere in this path (AGPL-free, commercially safe).

**Model:** PP-LiteSeg-T (STDC1 backbone + detail/attention decoder) from PaddleSeg (Apache-2.0), binary background-vs-body at 512px, transfer-learning from the stock PP-LiteSeg pretrained weights. Best checkpoint is selected by validation mIoU (`--do_eval`).

**Any body color:** the train split is extended in place with offline blue/green/random recolors of the exact SAM masks (masks unchanged, valid/test stay real), so brown, blue, green and gray bodies are all detected. A hue-bucketed test report (IoU per body color) is written at the end.

Outputs in `MyDrive/resistor_model/segmentation-pp-liteseg`: PaddleSeg output dir, exported inference model, ONNX, best-effort TFLite, test report. PaddleSeg `iter_*` checkpoints resume automatically.
